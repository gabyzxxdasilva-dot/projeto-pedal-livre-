Machine Learning - Regressão Linear

```python
from sklearn.linear_model import Lasso

#Atribuitos / Variáveis de entrada
orcamento = [[2000],[3250],[4700],[5450],[6421]]
#Alvo (target) / variável prevista
lucro = [3250,6500,6700,7000,7540]

#instalando modelo
modelo = Lasso()

#treinando modelo
modelo.fit(X=orcamento, y = lucro)

orcamento_previsto = 5000
previsao_lucro = modelo.predict([[orcamento_previsto ]])
print(f"com o orçamento de: R$ {orcamento_previsto:,.2f} o lucro {previsao_lucro[0]:,.2f}")
```
